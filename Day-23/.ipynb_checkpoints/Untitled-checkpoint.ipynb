# File Handling

```python
import os 
os.getcwd()
```

```python
#open("file_name","Mode").
f = open("f1.txt","r")

for line in f:
    print(line)

f.close() # closing the file, deleting from RAM
```

```python
#read(): to get all lines from the file

with open("f1.txt","r") as f:
    print(f.read()) # read the entire file
```

```python
#readlines(): to get all lines  into a list , to access particular line data.

with open("f1.txt","r") as f:
    lines = f.readlines()
    print(lines) # read the entire file
```

```python
#readlines(): to get all lines  into a list , to access particular line data.

with open("f1.txt","r") as f:
    lines = f.readlines()
    print(lines[4]) # read the entire file
```

```python
#readlines(): to get all lines  into a list , to access particular line data.

with open("f1.txt","r") as f:
    lines = f.readlines()
    print(*lines) #unpacking of complete list
```

# Write Functions

```python
# To write the concept inside the file
# We will use the "W" mode with write() fucntion
# if file don't exist then "w" mode will create a new file.
# read mdoe "r" don't create the file
```

```python
with open("Mallikharjuna.txt","w") as f:
    f.write("Mallikharjuna see his X daily in his dreams and get afraid")
```

```python
with open("Mallikharjuna.txt","r") as f:
    print(f.read())
```

```python
#In "W" mode data get overwrite on previous data
with open("Mallikharjuna.txt","w") as f:
    f.write("Mallikharjuna best friend is poorneswar, he always keep disturbing him.")
    
```

```python
with open("Mallikharjuna.txt","r") as f:
    print(f.read())
```

# append Mode:a

```python
with open("Mallikharjuna.txt","a") as f:
    f.write("Pawan kalyan is very innocent student.")
    
```

```python
with open("Mallikharjuna.txt","r") as f:
    print(f.read())
```

# write lines: Writing multiple lines together in the file

```python
with open("Mallikharjuna.txt","a") as f:
    f.writelines(["Poorneswar have 4 GFs. \n","but still he is alive"])
    
```

```python
with open("Mallikharjuna.txt","r") as f:
    print(f.read())
```

```python
# w+ mode: write/read mode
# w+ mode will clear the previous content
# W+ create new file if file don't exist

with open("sample.txt", "w+") as f:
    f.write(f"This is new file")
    f.seek(0)
    print(f.read())
```

```python
# r+ : read/wrie mode
# file must exist, otherwise shows error

with open("sample.txt", "r+") as f:
    f.write("mallikharjuna is very hard working")
    f.seek(0)
    print(f.read())
```
