# Polymorphism

1. The word polymorphism means "Many forms"
2. In polymorphism, some function or same operator can perform many different task.


## Types of Polymorphism

1. Function overloading
2. operator overloading
3. Method overloading


### 1.Function overloading

#These type function overloading is supported in c++ & Java.
#Based on the number of arguments passed to function, C++ and java can differentiate the functions.


def fun(var1):
    print(var1)
def fun(var1,var2):

fun(10)

### Using default arguments for function overloading

```python
def fun(var1, var2 = None,var3 = None):
    if var2 and var3 is not None:
        print(var1,var2,var3)
    else:
        print(var1)
#fun(10)
#fun(10,20)
fun(10,20,30)
```

```python
#Example:2

def calculator(a=0,b=0,c=0,d=0):
    return a+b+c+d
print(calculator(1889209,58378982))
print(calculator(100,200,300))
print(calculator())
```

```python
## *args for function for function overloading

#Example:2

def calculator(*args):
    return sum(args)
print(calculator(1889209,58378982))
print(calculator(100,200,300))
print(calculator())
```

```python
# Function overloading using decorator

from multipledispatch import dispatch

@dispatch(int)
def fun(var1):
    print(var1)

@dispatch(int,int)
def fun(var1,var2):
    print(var1,var2)

@dispatch(float,float,int,)
def fun(var1,var2,var3):
    print(var1,var2,var3)
fun(10)
fun(20,30)
fun(10.0,20.2,33)
```

```python
class Dosa:
    def __init__(self,variety,price):
        self.Variety = variety
        self.Price = price
    def __add__(self, obj1_val):
        total_price = self.Price + obj1_val.Price
        return total_price
    def __sub__(self, obj1_val):
        total_price = self.Price - obj1_val.Price
        return total_price
    def __mul__(self, obj1_val):
        total_price = self.Price * obj1_val.Price
        return total_price

    def __truediv__(self, obj1_val):
        total_price = self.Price / obj1_val.Price
        return total_price
    def __floordiv__(self, obj1_val):
        total_price = self.Price // obj1_val.Price
        return total_price
    def __mod__(self, obj1_val):
        total_price = self.Price % obj1_val.Price
        return total_price
        '''
        def __pow__(self, obj1_val):
        total_price = self.Price ** obj1_val.Price
        return total_price
        '''
    def __eq__(self, obj1_val):
        total_price = self.Price == obj1_val.Price
        return total_price
    def __gt__(self, obj1_val):
        total_price = self.Price > obj1_val.Price
        return total_price

    def __lt__(self, obj1_val):
        total_price = self.Price < obj1_val.Price
        return total_price



obj1 = Dosa("Egg Dosa", 80)
obj2 = Dosa("Onion Dosa",50)
#obj3 = Dosa("Ice-cream",100)

print(obj1 + obj2) # __add__
print(obj1 - obj2) #__sub__
print(obj1 * obj2) #__mul__
print(obj1 / obj2) #__truediv__
print(obj1 // obj2) #__floordiv__
print(obj1 % obj2)  # __mod__
#print(obj1 ** obj2) #__pow__

print(obj1 > obj2) # __gt__()
print(obj1 < obj2) #__lf__()
print(obj1 == obj2) #__eq__()


```

# Data Abstraction

->def: Abstraction hides internal implementation details and shows only essential features.

->why use data abstraction:

1. To hide the complex details
2. this make the code more readable and secure
-> libraries needed ABC, abstraction
   from abc import ABC, abstractmethod
   
-> Rules of data abstraction in python:
   1. Abstract class can have both abstract and concrete method.
   2. You cannot instance a object from abstract clss.
   3. All subclass must implement all abstract methods.

->Abstract method: A method declared but not defined/implemented/without code.
                 : abstract method is defined using @abstractmethod decorator
->Concrete method: This method have the body parts



```python
# Example 1:

from abc import ABC, abstractmethod

#abstract method
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self): #abstact method
        pass
    @abstractmethod
    def led_screen(self):
        pass
    
    def insurance(self): # concrete
        print("Please take Mahindra company insurance")
        


class hyd_mahindra(Head_Mahindra):
     def air_bag(self): 
        print("I will use Nylon to design the Air bag")

     def led_screen(self):
         print("i will use Sony LED")

class Noida_mahindra(Head_Mahindra):
    def air_bag(self):
        print("I will use Cotton to design the Air bag")
    def led_screen(self):
        print("i will use samsung")
        
obj_hyd = hyd_mahindra() # create object of hyd child class
obj_Noida = Noida_mahindra() # create object of Noida child class    
#obj_head = Head_Mahindra() #Can't instantiate abstract class Head_Mahindra

obj_hyd.air_bag()
obj_hyd.insurance()
```

```python
from abc import ABC, abstractmethod

# abstract class
class Head_Mahindra(ABC):
    @abstractmethod
    def air_bag(self):  # abstract method
        pass

    @abstractmethod
    def LED_Screen(self):
        pass

    def insurance(self):  # concrete
        print("Please take Mahindra company insurance")

class hyd_mahindra(Head_Mahindra): # child class
    def air_bag(self):  # abstract method
        print("I will use Nylon to design the Air bag")

    def LED_Screen(self):
        print("Sony LED")

class noida_mahindra(Head_Mahindra):
    def air_bag(self):  # abstract method
        print("I will use Cotton to design the Air bag")

    def LED_Screen(self):
        print("Samsung LED")

obj_hyd = hyd_mahindra()   # create object of hyd child class
obj_noida = noida_mahindra()

obj_hyd.insurance()
# obj_head = Head_Mahindra()
```
