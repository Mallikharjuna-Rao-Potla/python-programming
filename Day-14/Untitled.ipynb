# Insert, delete and update operations

```python
tuple1 = (10,20.5,"Hello",[10,20,30])
```

```python
tuple1[2]
```

```python
tuple1[3][1]
```

```python
tuple1[2][3]
```

```python
tuple1[3][1] = 400
```

```python
tuple1
```

1. Tuple items cannot be updated directly.
2. Completed item in tuple cannot be updated directly.
3. But if item is mutable then elements inside this mutable item can be updated.

```python
tuple3 = (10,20,30,40,50)
list1 = list(tuple3)
print(list1)
list1.append(11)
print(list1)
tuple3  = tuple(list1)
print(tuple3)
```

```python
list1 = list(tuple3)
print(list1)
list1.remove(10)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

```python
list1 = list(tuple3)
print(list1)
list1.pop()
print(list1)
list1.insert(4,3000)
print(list1)
tuple3 = tuple(list1)
print(tuple3)
```

#### popular inbuilt function for tuple

```python
salary = (6,70,100,90,10,80,120)
print(len(salary))
print(min(salary))
print(max(salary))
print(sum(salary))

sorted_list = sorted(salary)
print(sorted_list)
```

1. sort() is inplace function
2. inplace means sorting is done in the same data structure
3. sorted() function return the list
4. sort() function didn't return list.
