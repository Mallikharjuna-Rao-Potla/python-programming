# Recursive function

1. Def : a function calling itself directly or in-directly
2. Termination condition is MUST in recursion
3. In the recursive call,we must change the parameter value.


```python
def fun(i):
    if i < 1:
        return
    print(i)
    fun(i-1)
fun(6)
```

```python
def add(a,b):
    print(a + b)
res = add(10,20)
print(res)
```

```python
def countdown(n):
    if n < 0:
        print("Done")
        return 
    #print(n)
    countdown(n-1)
    print(n)

countdown(10)


```

# Enumerate functions


# enumerate() function add the counter(index) with the list items.


```python
fruits_names  = ["mamidipandu","panasakaya","jamapandu"]

for name in fruits_names:
    print(name)
```

```python
fruits_names  = ["mamidipandu","panasakaya","jamapandu"]

for index,name in enumerate(fruits_names,start=1240):
    print(index,name)
```

# Zip-Function()

```python
## zip function is used to combine/zip multiple list/tuple items,element-wise.
```

```python
Hero = ["Mahesh","Mallikharjun","Pavan kalyan","Balaji"]
Heroines = ["anushka","sam","rashmika","kajal"]
Movie_name = ["Khaleja","No_movie","maghadeera","panja"]

pairs = zip(Hero,Heroines,Movie_name)
print(list(pairs))
```

### Decorators

###### Def: a decorator is a function that makes another function as input, adds extra functionality and returns a modified fucntion,without changing the origional function code.

```python
def decorator_fun(func):
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients")
        func()
        print("add cherry") # extra functionality
        print("add chocolate")
    return wrapper_fun
def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")

decorator_function = decorator_fun(cake)
decorator_function()
```

```python
# example-2:use of @ to represent the decorators
def decorator_fun(func):
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients")
        func()
        print("add cherry") # extra functionality
        print("add chocolate")
    return wrapper_fun
    
@decorator_fun

def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```

```python
#mulitple decorate for same function
def decorator_fun1(func):
    def wrapper_fun(): # wrapper function add the extra functionality
        print("Buy ingredients")
        func()
    return wrapper_fun

def decorator_fun2(func):
    def wrapper_fun(): # wrapper function add the extra functionality
        func()
        print("add cherry") # extra functionality
        print("add chocolate")
    return wrapper_fun
    
@decorator_fun1
@decorator_fun2

def cake():
    print("Take the flour")
    print("add eggs and other ingredient")
    print("put the cake in oven")
cake()
```
