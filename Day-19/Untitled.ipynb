```python
# Example 2: constructor function
#Def: constructor function automatically get callled when object is created.

# class variable vlaue is same for all the objects.

# self is as variable which points to the current objects

class Bike:
    No_tyre = 2 # class variable
    def __init__(self,name,price):
        self.bike_name = name # instance varible 
        self.bike_price = price
        print(self.bike_name,self.bike_price)
     

obj1 = Bike("Yamaha",100000)
obj2 = Bike("Hero",92000)
```

```python
# Example 3 : 
# member function don't get called automatically, we need to call 

class Bike:
    No_tyre = 2 # class variable
    def __init__(self,name,price):
        self.bike_name = name # instance varible 
        self.bike_price = price
        print(self.bike_name,self.bike_price)

    def display(self):
        print("I am a member function")

obj1 = Bike("Yamaha",100000)
obj2 = Bike("Hero",92000)

obj1.display()
obj2.display()
```

# Encapsulation


```python
# Def : Encapsulation refers to the binding of data(attributes) and functions into one unit(class).

# Encapsulation restrict the direct access of the data.

# we can access the variables/data/attribute through object outside the class.
# we can access the variables/data/attribute through fucntions defined inside the class.
 

class Bike:
    No_tyre = 2 # class variable
    def __init__(self,name,price):
        self.bike_name = name # instance varible 
        self.bike_price = price
        

    def display(self):
        print(self.No_tyre,self.bike_name,self.bike_price)

obj1 = Bike("Yamaha",100000)
obj1.display()
print(obj1.No_tyre)
print(obj1.bike_name)
print(obj1.bike_price)
```

### Access specifers / type of attributes in Encapsulation

1. Public
2. protected
3. private


```python
class marvels:
    def __init__(self, name,power,weapon):
        self.Name = name   #public
        self._Power = power #protected
        self.__Weapon = weapon #private

    def show(self):
        print(self.Name,self._Power,self.__Weapon)

obj1 = marvels("Thor",1000,"Hammer")
obj1.show()

print(obj1.Name)
print(obj1._Power)

# Protected variables can be access outside the class, but you should not access.
# In python public,protected,private concpet is used just for sign/warning

print(obj1._marvels__Weapon) # obj._classname__privateVariable


        
```

# Inheritance 

```python
# Def: Inheritance allows us to define a classs that inherits the properties or functions of the parent class
```

```python
# parent class = Base class
# child class = derived class
```

# Types of inheritance :
1. Single inheritance : one parent , one child
2. Multiple inheritance : Many parents one child class
3. multilevel inheritance : son -> Father -> grand father
4. Hierarichal inheritance : One parent Many child
5. Hybrid inheritance : many parent many child. combine any two or more inheritance.

```python
# 1.Single inheritance : one parent , one child
```

```python
class hyd_food:
    taste = "very very spicy"

    def  Biryani(self):
        print("Very very tasty Birayani")
class PistaHouse(hyd_food):
    pass
    
obj_child = PistaHouse() #it is the object of the child class

# child class object can acess the properties & function of parent class
print(obj_child.taste)
obj_child.Biryani()
```

```python
#Example 2: super() function

#Super() : used to access the parent properties and fucntion from inside of the child class

class hyd_food:
    taste = "very very spicy"

    def  Biryani(self):
        print("Very very tasty Birayani")
#child class        
class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)
    
obj_child = PistaHouse() #it is the object of the child class
obj_child.display()
```
