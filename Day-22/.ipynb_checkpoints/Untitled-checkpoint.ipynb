# Exception Handling

```python
# Def : Exception allows to respond to error without crashing the running application.
# with EH , we can give our own error message.
```

#### Types of blocks in Exception Handling:
1. try : put that code where error can occur 
2. except : handle the erro
3. else : put code which you want to execute when their is NO error
4. finally : execute either error or No error
   

```python
print(x)

print("Hello world")

a,b = 10,20
print(a+b)
```

```python
# Example:1 stucture of try

try:
    print(x)
except:
    print("Some Error occur")

print("Hello world")
a,b = 10,20
print(a+b)
```

```python
try:
    # Code that might raise an exception
    user_input = int(input("Enter a number: "))
    result = 100 / user_input
    print(f"100 divided by your number is: {result}")
except:
    # Code that executes if an exception occurs
    print("Something went wrong! Please try again.")
```

```python
# Example:

try:
    x = 10
    print(x)
except:
    print("Some Error occur")

print("Hello world")
a,b = 10,20
print(a+b)
```

```python
# Example:2 Exception as e

try:
    print(y)
except Exception as e:
    print(f"Complier error:{e}")
    print(type(e))

print("Hello world")
a,b = 10,20
print(a+b)
```

```python
# Example:3 else block
try:
    y = 10
    print(y)
except Exception as e:
    print(f"complier error : {e}")
    print(type(e))
else:
    print("Hurry, we successfully excuted code")

print("Hello world")
a,b = 10,20
print(a+b)
```

```python
# example:4 finally block: execute if NO error + Error

try:
    y = 20
    print(y)
except Exception as e:
    print(f"complier error:{e}")
    print(type(e))
else:
    print("Hurry, we successfully excuted code")
finally:
    print("I will execute whatever happen")
```

```python
# example:4 finally block: execute if NO error + Error

try:
    print(m)
except Exception as e:
    print(f"complier error:{e}")
    print(type(e))
else:
    print("Hurry, we successfully excuted code")
finally:
    print("I will execute whatever happen")
```

```python
try:
    num = int("abc")
    print(num)
except NameError:
    print("This is NameError")
except ZeroDivisionError:
    print("This is ZeroDivisionError")
except IndexError:
    print("This is index error")
except KeyError:
    print("This is keyerror")
except ValueError:
    print("This is valueerror")
except Exception as e:
    print(type(e),e)
```

### use of raise Keyword

```python
# raise is used to generate user define error

age = -22

if age < 0:
    raise Exception("Sorry,Age cannot be in negative")

```

```python
def process_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative")
    if age > 120:
        raise ValueError("Are you sure this age is correct?")
    return f"Processing age: {age}"

try:
    print(process_age(-15))
except ValueError as error:
    print(f"Invalid age: {error}")
```

# Creating Custom Exceptions

```python
class InsufficientFundsError(Exception):
    """Raised when a transaction would result in a negative balance"""
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientFundsError(f"You tried to withdraw ${amount} but your balance is only ${balance}")
    return balance - amount

try:#new_balance = withdraw(100, 150)
    new_balance = withdraw(20,100)
except InsufficientFundsError as e:
    print(f"Transaction failed: {e}")
```
