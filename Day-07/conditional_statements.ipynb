# Conditional Statements in Python

```python
# Example-1: Structure of if statement
a = 10
if (a % 2 == 0):
    print(a)
```

```python
if a%2 == 0: print(a)
```

```python
#Example 2: if-else condition
a = 1243
if a%2 == 0:
    print("Even")
else:
    print("odd")
```

```python
a = int(input())
if a%3 == 0:
    print("Hello")
else:
    print("world")
```

```python
# Example-3 : elif condition

Biryani = input("Famous hyderabad birayani")
if Biryani == "paradise":
    print("Spicy & Tasty")
elif Biryani == "shahdab":
    print("Double mass")
elif Biryani == "road side":
    print("Budget friendly")
else:
    print("PG Biryani")
```

```python
#replaces elif with if-else
Biryani = input("Famous hyderabad birayani:")
if Biryani == "paradise":
    print("Spicy & Tasty")
    if Biryani == "shahdab":
        print("Double mass")
    if Biryani == "road side":
        print("Budget friendly")
else:
    print("PG Biryani")
```

```python
# Ternary 
Biryani = input("Famous hyderabad birayani:")
print("Spicy & Tasty") if Biryani == "paradise" else print("Double masala") if Biryani == "shahdab" else print("Budget friendly") if Biryani == "road side" else print("PG birayani")
```
