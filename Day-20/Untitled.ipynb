# Inheritance 

```python
# Def: Inheritance allows us to define a classs that inherits the properties or functions of the parent class
```

```python
# parent class = Base class
# child class = derived class
```

# Types of inheritance :
1. Single inheritance : one parent , one child
2. Multiple inheritance : Many parents one child class
3. multilevel inheritance : son -> Father -> grand father
4. Hierarichal inheritance : One parent Many child
5. Hybrid inheritance : many parent many child. combine any two or more inheritance.

# 1.Single inheritance : one parent , one child

```python
class hyd_food:
    taste = "very very spicy"

    def  Biryani(self):
        print("Very very tasty Birayani")
class PistaHouse(hyd_food):
    pass
    
obj_child = PistaHouse() #it is the object of the child class

# child class object can acess the properties & function of parent class
print(obj_child.taste)
obj_child.Biryani()
```

```python
#Example 2: super() function

#Super() : used to access the parent properties and fucntion from inside of the child class

class hyd_food:
    taste = "very very spicy"

    def  Biryani(self):
        print("Very very tasty Birayani")
#child class        
class PistaHouse(hyd_food):
    def display(self):
        super().Biryani()
        print(super().taste)
    
obj_child = PistaHouse() #it is the object of the child class
obj_child.display()
```

# 2.Multiple inheritance : Many parents one child class

```python
class A:
    def display(self):
        print("I am parent class ")
class B:
    def show(self):
        print("I am also parent class")
class c(A,B):
    pass
obj_c = c()
obj_c.display()
obj_c.show()
```

```python
class A:
    def display(self):
        print("I am parent class ")
class B:
    def display(self):
        print("I am also parent class")
class c(A,B): # sequence in which parent are assign that matters.
    pass
obj_c = c()
obj_c.display()
obj_c.display()
```

```python
#Example :3 constructor
class A:
    '''
    def __init__(self):
        print("I am class A constructor")
        '''
    
class B:
    def __init__(self):
        print("I am class B constructor")
   
class c(A,B):
    '''
    def __init__(self):
        print("Iam child class constructor")
        '''

obj_c = c()
```

```python
# Example-2:super()
# MRO : Method Resolution Order

class A:
    def display(self):
        print("I am parent class ")
        super().display()
class B:
    def display(self):
        print("I am also parent class")
class c(A,B):
    def hello(self):
        super().display()
obj_c = c()
obj_c.hello()

# MRO : Method Resolution Order
print([i.__name__ for i in c.__mro__])
```

# 5. Hybrid inheritance

```python
# Example-1:structure of hybrid inheritance
class A:
    def __init__(self):
        print("I am class A conbstructor fucntion ")
        
    A_var = "A class"
    
class B(A):
      def __init__(self):
        print("I am class B conbstructor fucntion ")
      B_var = "B class"

class C(B):
      def __init__(self):
        print("I am class C conbstructor fucntion ")
      C_var = "C class"
class D(B):
      def __init__(self):
        print("I am class D conbstructor fucntion ")
      D_var = "D class"
class E(C,D):
      def __init__(self):
        print("I am class E conbstructor fucntion ")
      E_var = "E_var"

obj_E = E()
print(obj_E.A_var)
```
