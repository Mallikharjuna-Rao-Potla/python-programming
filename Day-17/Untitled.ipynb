## Default Parameters

```python
#assign a default value to the "var" variable
def fun(var = "Nothing"):
    print(var)
fun()
fun(100)
```

```python
def fun2(x=10,y=20):
    print(x,y)
fun2()
```

```python
def fun2(x=10,y=20):
    print(x+y)
fun2()
fun2 (100,200)
```

## passing arbitrary number of arguments using *args

```python
def fun(*args):
    for i in args:
        print(i)
fun()
fun("chinki")
fun("chinki","minki")
```

```python
def fun(*args):
    for i in args:
        print(i)
fun()
fun(10,20)
fun(100,200,300)
```

```python
def fun(*args):
    for i in args:
        return i
fun()
fun(10,20)
fun(100,200,300)
```

### passing key-value based arguments using **kargs

```python
def pavan(**GF):
    for key in GF:
        print(key, " : ",GF[key])
pavan(chinki = 150, minki = 100 )
pavan(chinki = 150)
pavan()
        
```

## Lambda functions

1. Def : a lambda function is a single line of code that is used to perform
a short task without defining a full function with def.

2. lambda function don't have any name.

3. Although lambda function can receive arguments and process them.

4. lambda function are popular used inside other function, to perform small function




```python
obj = lambda a : a + 100
print(obj(10))
```

```python
obj = lambda a,b : a + b
print(obj(100,200))
```

```python
obj = lambda a,b ,c : a * b * c
print(obj(10,2,3))
```

```python
# using lambda functions inside the other function
def fun(p):
    return lambda a : a**p
  
obj = fun(10)

print(obj(2))
print(obj(3))
print(obj(4))
```

```python
# using lambda functions inside the other function
def fun(p):
    obj =  lambda a : a**p if a%2 == 0 else a+p
    return obj
    
  
obj = fun(2)

print(obj(2))
print(obj(3))
print(obj(4))
```

```python
# sort the below given list of tuples

students = [("pavan",80),("dinesh",75),("chandra",78)]
sorted_list = sorted(students, key = lambda a : a[1])
print(sorted_list)
```

```python
# sort the below given list of tuples

students = [("pavan",80,5),("dinesh",75,10),("chandra",78,15)]
sorted_list = sorted(students, key = lambda a : a[2])
print(sorted_list)
```

```python
#using the dictionary
students = {"pavan":80,"dinesh":75,"chandra":78}
data = students.items()
print(data)
sorted_list = sorted(data, key = lambda a : a[1])
print(sorted_list)
```
