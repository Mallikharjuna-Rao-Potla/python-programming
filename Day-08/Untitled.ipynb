# Match case


#### Match case is used to replace multiple if else condition
#### Match case is more powerful than the Switch case statment
#### Match case works with variable and data structures

```python
#Example:1
num = int(input("Enter a number:"))
match num:
    case 10:
        print("You entered the Number 10")
    case 20:
        print("You entered the number 20")
    case 30:
        print("You entered the number 30")
    case _:
        print("something else")
```

```python
Biryani = input("Famous hyderabad birayani:")
match Biryani:
    case "paradise":
        print("You ordered the paradise birayani")
    case "shahdab":
        print("You ordered the shahdab birayani")
    case "road_side":
        print("You looking for budget friendly")
    case _:
        print("Go & Eat PG birayani")
```

```python
# Example-2 passing data structure to match case

data = [1,2,3,4,5]
match data:
    case [5,4,3,2,1]:
        print("Hello")
    case [1,2,3,4,5]:
        print("world")
```

```python
# Example-3: if condition with match case
val = int(input("Enter an number"))
match val:
    case x if x > 0:
        print("it is a postive number")
    case y if y < 0:
        print("it is a negative number")
    case _:
        print("it is a zero")
```
