##### Basic interview Question: string is mutable or immutable?
##### Ans: string is immutable
#### Mutable : insert, delete and update operation can be performed
#### immutable : insert , delete and update operation can't be performed




### Operations on strings

1. concatenation (+)
2. Repetition (*)
3. Membership (in,not in)
4. Indexing and slicing

#### 1. Concatenation

```python
str1 = "Babai"
str2 = "likes"
str3 = "pinni"
print(str1 + " "+ str2 + " "+ str3)
```

#### 2.Repetition(*)

```python
str = "bla "

print(10 * str)
```

#### 4. Membership(in, not in)

```python
str1 = "Helloworld"

print("Hell" in str1)
print("Helll" in str1)
print("o w" in str1)
print("Helo" not in str1)
```

```python
print(len(str1))
```

#### 5.Indexing and Slicing

```python
# string[start:end:step]
Heroine = "samantha"

print(Heroine[4])
print(Heroine[-6])
print(Heroine[0:5])
print(Heroine[0:5:3])
print(Heroine[0:5:2])

print(Heroine[::])
print(Heroine[::2])
print(Heroine[7:1:1]) # if step size is +ve number then it's return nothing.because +ve no. moving always right side
print(Heroine[6:3:-1])
print(Heroine[-1])
'''
print(Heroine[:5:3])
print(Heroine[:-4])
print(Heroine[-1::-2])
'''
```

```python
print(Heroine[-6])
```

```python
print(Heroine[-1:-6]) #output : empty because the step is +ve number taking right side 
```

```python
print(Heroine[-1:-6:-1])
```
