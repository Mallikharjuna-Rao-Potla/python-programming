# for loop

```python
animals = ["Puli","Pilli","Kukka","pandi","kothi"]
for i in animals:
    if i == "Kukka":
        idx = animals.index(i)
        print(idx)
    print(i)
```

```python
animals = ["Puli","Pilli","Kukka","pandi","kothi"]
for i in animals[:2]:
    print(i)
```

```python
food_name = "chicken_Dum_Masala_Biryani"
for i in food_name:
    if i == "k":
        print("I found k")
    print(i)
```

```python
# for loop with range()
# syntax of range:range(start,end,step)
# syntax of range:range(start,end,step)
# start = 0
# end = end - 1
# step = 1
for i in range(10):
    print(i)
```

```python
for i in range(4,10):
    print(i)
```

```python
for i in range(4,10,3):
    print(i)
```

```python
# Example 4: break statement
# break statement will bring the control out of the loop
for num in range(1,6):
    if num == 3:
        break
    print(num)
```

```python
# Example-5: continue statement
# continue statement take the control to the start of the loop
for num in range(1,6):
    if num == 3:
        continue
    print(num)
```

```python
# Example - 6: nested loop
# for every value of outer loop,inner loop will completely executed
for i in range(1,4):
    for j in range(1,3):
        print(i,j)
```

```python
# Example - 6: nested loop
for i in range(3,6):
    for j in range(2,7):
        print(i,j)
#Time complexity : O(n * m)
```

```python
# Example-7: else statement
# else statement executes if loop complete execution successfully

for i in range(5):
    if i == 2:
        continue
    elif i == 4:
        break
    else:
        print(i)
    pass
else:
    print("loop executes successfully")
```

### Patterns

```python
for i in range(5):
    for j in range(5):
        print("*",end = "")
    print()
```

```python
for i in range(1,5):
    for j in range(i):
        print("*", end = "")
    print()
```

```python
for i in range(5,0,-1):
    for j in range(i):
        print(" * ", end = "")
    print()
```

```python
n = 6
for i in range(n,0,-1):
    print(" " * i + "* " * (n-i+1))
```
