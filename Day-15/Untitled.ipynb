#### delete operation on set(remove(),discard(),pop())

```python
dinner_set = {"Bagara rice","Birayani","papu","chapati","fish","mutton","prawns"}

#remove() function delete the one item whose name is provided
#remove() will raise an error if item don't exist
dinner_set.remove("prawns")
print(dinner_set)
```

```python
#discard() function delete the one item whose name is provided
#disard DO NOT raise error if item is not present
dinner_set.discard("Birayani")
print(dinner_set)
```

```python
#pop() will delete any random item
dinner_set.pop()
print(dinner_set)
```

```python
dinner_set.clear()
print(dinner_set)
```
