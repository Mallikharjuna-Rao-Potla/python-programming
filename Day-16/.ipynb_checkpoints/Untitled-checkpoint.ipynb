#inteview Questions asked in the interview:

Q1.Dictionary is mutable or immutable?
Ans:dictionary is mutable(insert,delete,update(values))
But keys are immutable because the keys are not updated.

Q2. How much time take to insert , update ,delete in dictionary?
Ans: Best case, Average case : O(n) ,  worst case : O(n).

Q3. Duplicate items are allowed or not ?
Ans:values are allowed ,but keys values are not allowed it can be overriiden.

Q4.Dictionary is ordered or unordered?
Ans: Before python 3.7 version dictionary was unordered.
to make the access of elements faster

```python
#Q1.Dictionary is mutable or immutable?
int_dict = {1:'mango',2:'banana',3:'orange'}
print(int_dict)

float_dict = {1.1 : 'mango',2.2 : 'Banana',3.3:'orange'}
print(float_dict)

str_dict = {'a' : 'mango','b' : 'Banana','c':'orange'}
print(str_dict)
```

```python
hetro_dict = {'a': 'mango',2:'banana',3.3  : 'orange'}
print(hetro_dict)
```

```python
tuple_dict = {(10,30):'low',(35,60):"averge",(70,90):"high"}
print(tuple_dict)
```

```python
list_dict = {[10,30]:'low',[35,60]:"averge",[70,90]:"high"}
print(tuple_dict)
```

### Accessing dictionary items

```python
drink_items = {
    "Thupsup" : 50,
    "sugarcane" : 20,
    "coconutwater" : 60,
    "water": 20,
    "rose milk" : 30,
}


val = drink_items["sugarcane"]
print(val)
```

```python
val2 = drink_items.get("water")
print(val2)
```

```python
#val = drink_items["lemon"]
#print(val)

val2 = drink_items.get("lemon","key not available")
print(val2)

val2 = drink_items.get("coconutwater","key not available")
print(val2)
```

```python
for i in drink_items: #i carries the key values
    print(i,drink_items[i]) 
```

```python
print(drink_items.keys())
```

```python
print(drink_items.values())
```

```python
# nested dictionary
nested_dict = {
    100 : {"Hello":10,"world": 20},
    200 : {"A":50,"B":60}
}
nested_dict[100]["world"]
```

### inserting & updating operation on dictioary

```python
x_dict = {"name": "chinki","age":80}

x_dict["age"] = 180  #update
print(x_dict)

x_dict['city'] = "kurnool" #insert
print(x_dict)

#insert and update using the update() fuction
#update() can insert and update multiple together

x_dict.update({'age':18,'occupation':'IAS',"name":"minki","salary": 1.5})
print(x_dict)
```

### Delete the items from dictionary

```python
# pop(),popitem(), clear(),del()

x_dict
```

```python
#pop removes the item whose key is specified
x_dict.pop('salary')
print(x_dict)
```

```python
#popitem removes the last item present in the dict.
x_dict.popitem()
print(x_dict)
```

```python
#clear deletes all items present in the dict and return the empty dict
x_dict.clear()
print(x_dict)
```

```python
#del deletes the  entire delete even empty_dict also
del x_dict
```
