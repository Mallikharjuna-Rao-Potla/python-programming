# Interview Question from the list.
1. List is mutable or immutable data structures?
Ans: Yes, list is mutable(insert, update, delete can be performed)

2. what is the time complexity to insert one element in list?
Ans: Avg.case: O(1),wc = o(n)

3. pthon complier reserve some space for list in RAM 

4. if this space get full, space get double by the python complier.

5. List is Dynamic or static data structure?
Ans:List is Dynamic data structure means,
Means it can reserve the space at Run time.

6. Do list support Heterogenous elements?
Ans:Yes

7. Do list support duplicate elements?
Ans:Yes

8. List is ordered or not?

Ans: Yes


```python
list1 = [1,2,3,3,4.5,"Hello",True]
print(list1)
print(list1[5])
#print(sys.getsizeof(list1))
```

### Creating different types of list

```python
empty_list = []
empty_list2 = list()
print(empty_list, type(empty_list))
print(empty_list2, type(empty_list2))
```

```python
Het_list = [10,20.5,"Hello",True]
print(Het_list)
```

```python
dup_list = [10,10,20,30,40]
print(dup_list)
```

```python
# nested list 
nested_list = [[1,2],[10,20],["Hello","World"],1000]
print(nested_list)
```

```python
len(nested_list)
```

```python
print(nested_list[3])
```

```python
nested_list[1][1]
```

#### List concatenation

```python
concat_list = Het_list + dup_list + nested_list
print(concat_list)
```

#### Accessing List items

```python
#start = 0, end = last element of the list, step = 1
Hours_list = [1,5,4,2,3,4,2,3,1.5,2,2,3,20,24]

```

```python
#start = 0, end = last element of the list, step = 1
Hours_list[4:-14:-1]
```
