# Loops in Python

## While Loop

```python
i = 1
while(i <= 10):
    print(i)
    i += 1
print(i)
```

```python
i = 1
while(i <= 10):
    i += 1
    print(i)
```

## Break Statement

```python
#Example:2 break statement
#break statement helps to come out of the loop
i = 1
while(i <= 10):
    print(i)
    if i == 5:
        break
    i += 1
```

```python
i = 1
while(i <= 10):
        i += 1
        print(i)
        if i == 5:
            break
    
```

# contine statement
1. Continue will take u control to the loop / start of the loop.

```python
i = 1
while i < 10:
    i += 1
    if i < 5:
        continue
    print(i)
```

```python
i = 1
while i < 10:
    i += 1
    print(i)
    if i < 5:
        continue
```

```python
#  Exapmple 4 :else statement with while loop
# else statement executes only if loop is successfully completed.
i = 1
while i <= 10:
    print(i)
    i += 1
else:
    print("Loop is executed")    
```

```python
i = 1
while i <= 10:
    print(i)
    i += 1
    if i > 5:
        break
else:
    print("Loop is executed")  
```
